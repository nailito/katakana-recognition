```python
metadata = load_metadata()

df_train, df_valid, df_test = split(
    metadata,
    training=0.7,
    validation=0.15,
    test=0.15,
    seed=12,
)
```

```python
unique_char = sorted(metadata["char"].unique())
char_dic = {i : value for value, i in enumerate(unique_char)}
df_train["label"] = df_train["char"].map(char_dic)
df_valid["label"] = df_valid["char"].map(char_dic)
df_test["label"] = df_test["char"].map(char_dic)
```

```python
from src.input_pipeline import load_image
test_path = str(df_train["image_path"].iloc[0])
#load_image(test_path,0)
```

```python
paths = df_train["image_path"].astype(str).to_numpy()
labels = df_train["label"].to_numpy()

ds_train = tf.data.Dataset.from_tensor_slices((paths,labels))
ds_train_shuffled = ds_train.shuffle(buffer_size=len(df_train),seed=12)

train_ds = (ds_train_shuffled
            .map(load_image,num_parallel_calls=tf.data.AUTOTUNE)
            .batch(32)
            .prefetch(tf.data.AUTOTUNE))
```
