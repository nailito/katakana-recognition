# Archived: baseline CNN with fixed hyperparameters

This notebook establishes a reference experiment. The settings below are starting choices, not universally optimal values. Hyperparameter comparisons belong to a later experiment. The test set is reserved for final evaluation.

Run all cells in order to create and train a fresh model.

## 1. Imports

```python
#%load_ext autoreload
#%autoreload 2
import sys
from pathlib import Path
import matplotlib.pyplot as plt
import tensorflow as tf

keras = tf.keras
layers = keras.layers
Sequential = keras.Sequential

project_root = Path.cwd()
if project_root.name == "notebooks":
    project_root = project_root.parent

if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))

from src.load_metadata import load_metadata
from src.split_dataset import split
from src.input_pipeline import build_dataset
```

## 2. Fixed configuration

Adam uses a learning rate of 0.001. Keep the existing architecture and writer split for this baseline. Fixing the seed improves reproducibility but does not guarantee identical results on every platform.

```python
SEED = 12
BATCH_SIZE = 32
LEARNING_RATE = 0.001
FILTERS = (32, 64)
KERNEL_SIZE = 3
DENSE_UNITS = 128
MAX_EPOCHS = 10
PATIENCE = 3
INPUT_SHAPE = (63, 64, 1)

keras.utils.set_random_seed(SEED)
```

## 3. Data and shared labels

Pixels are already normalized by the input pipeline. Do not add another rescaling layer.

```python
metadata = load_metadata()

df_train, df_valid, df_test = split(
    metadata,
    training=0.7,
    validation=0.15,
    test=0.15,
    seed=SEED,
)

unique_char = sorted(metadata["char"].unique())
char_dic = {char: index for index, char in enumerate(unique_char)}
df_train["label"] = df_train["char"].map(char_dic)
df_valid["label"] = df_valid["char"].map(char_dic)
df_test["label"] = df_test["char"].map(char_dic)


train_ds = build_dataset(df_train, batch_size=BATCH_SIZE, training=True, seed=SEED)
val_ds = build_dataset(df_valid, batch_size=BATCH_SIZE, training=False)
```

## 4. Model

```python
keras.backend.clear_session()
keras.utils.set_random_seed(SEED)

model = keras.Sequential([
    layers.Input(shape=INPUT_SHAPE),
    layers.Conv2D(FILTERS[0], KERNEL_SIZE, activation="relu"),
    layers.MaxPooling2D(),
    layers.Conv2D(FILTERS[1], KERNEL_SIZE, activation="relu"),
    layers.Flatten(),
    layers.Dense(DENSE_UNITS, activation="relu"),
    layers.Dense(len(unique_char), activation="softmax"),
])

model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=LEARNING_RATE),
    loss=keras.losses.SparseCategoricalCrossentropy(),
    metrics=["accuracy"],
)
model.summary()
```

## 5. Check a batch without updating weights

```python
for images, labels in train_ds.take(1):
    predictions = model(images, training=False)
    print("Images:", images.shape)
    print("Labels:", labels.shape)
    print("Predictions:", predictions.shape)
    assert tuple(images.shape[1:]) == INPUT_SHAPE
    assert predictions.shape == (images.shape[0], len(unique_char))
```

## 6. Training

Stop after three consecutive epochs without an improvement in validation loss, and restore the weights from the best epoch. Ten epochs is the maximum, not a required training duration.

```python
early_stop = keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=PATIENCE,
    restore_best_weights=True,
)

history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=MAX_EPOCHS,
    callbacks=[early_stop],
    shuffle=False,  # Training examples are already shuffled by the input pipeline.
)
```

## 7. Training and validation loss

Compare both curves. A decreasing training loss alongside an increasing validation loss suggests overfitting. The dotted line marks the epoch selected by EarlyStopping. These curves describe the training history before the best weights are restored.

```python
epochs = range(1, len(history.history["loss"]) + 1)
best_epoch = min(epochs, key=lambda epoch: history.history["val_loss"][epoch - 1])

fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(epochs, history.history["loss"], label="Training")
ax.plot(epochs, history.history["val_loss"], label="Validation")
ax.axvline(best_epoch, linestyle=":", color="gray", label=f"Lowest val_loss: epoch {best_epoch}")
ax.set(xlabel="Epoch", ylabel="Cross-entropy loss", title="Baseline learning curves")
ax.set_xticks(list(epochs))
ax.legend()
ax.grid(alpha=0.2)
fig.tight_layout()
plt.show()
```

## Character recall and common confusions

Some Katakana are visually similar, such as ソ/ン and ワ/ウ. To assess whether the model distinguishes them, we examine both recall per character and the most frequent incorrect predictions.

Recall measures the proportion of a character's images that are correctly recognized. For the remaining images, we identify which characters the model predicts instead.

The table below shows the three most frequent confusions for each true character. It reports their frequency among all images of that character and their share of its errors. The error breakdown may therefore help identify systematic confusions hidden by overall accuracy.


```python
import numpy as np
import pandas as pd

val_ds = build_dataset(
    df_valid,
    batch_size=BATCH_SIZE,
    training=False,
)

probabilities = model.predict(val_ds)
predicted_labels = probabilities.argmax(axis=1)
true_labels = df_valid["label"].to_numpy()

assert len(predicted_labels) == len(true_labels)

characters = np.asarray(unique_char)

results = pd.DataFrame({
    "Actual": characters[true_labels],
    "Predicted": characters[predicted_labels],
})
results["Correct"] = results["Actual"] == results["Predicted"]

# Summary for each true character.
summary = results.groupby("Actual").agg(
    Support=("Correct", "size"),
    Recall=("Correct", "mean"),
)
summary["Errors"] = (
    summary["Support"] * (1 - summary["Recall"])
).round().astype(int)

confusions = (
    results.loc[~results["Correct"]]
    .groupby(["Actual", "Predicted"])
    .size()
    .rename("Count")
    .reset_index()
    .join(summary, on="Actual")
)

confusions["Rate among all examples"] = (
    confusions["Count"] / confusions["Support"]
)
confusions["Share of errors"] = (
    confusions["Count"] / confusions["Errors"]
)

top_confusions = (
    confusions
    .sort_values(
        ["Actual", "Count", "Predicted"],
        ascending=[True, False, True],
    )
    .groupby("Actual", sort=False)
    .head(3)
    .sort_values(
        ["Recall", "Actual", "Count"],
        ascending=[True, True, False],
    )
)

display(
    top_confusions[
        [
            "Actual",
            "Predicted",
            "Support",
            "Recall",
            "Count",
            "Rate among all examples",
            "Share of errors",
        ]
    ]
    .reset_index(drop=True)
    .style
    .format({
        "Recall": "{:.1%}",
        "Rate among all examples": "{:.1%}",
        "Share of errors": "{:.1%}",
    })
    .hide(axis="index")
)
```

## Visual inspection of common confusions

Compare four sampled errors for each selected pair with correctly recognized examples of both characters. Each figure contains three rows: errors, correct source-character predictions, and correct target-character predictions. Sampling uses a fixed seed for repeatable comparisons; these examples do not represent every error.

Look for differences in stroke orientation, position, spacing, and overall shape. The displayed score is the softmax probability of the predicted class, not a guarantee of correctness. Writer identifiers help trace examples.

Run the imports, configuration, data preparation, and confusion-analysis cells first. To reuse the saved baseline without training, load it with `model = keras.models.load_model(project_root / "models" / "katakana_baseline.keras")` before running the confusion analysis. Skip the model construction, training, and saving cells in that case.

If no Japanese font is available, images are still displayed without character titles; the compared characters are printed above each figure.

```python
from PIL import Image
from src.fonts import find_japanese_font

# results and probabilities must come from the same unshuffled val_ds.
assert len(results) == len(df_valid) == len(probabilities)
inspection = results.copy()
inspection["image_path"] = df_valid["image_path"].to_numpy()
inspection["writer_id"] = df_valid["writer_id"].to_numpy()
inspection["confidence"] = probabilities.max(axis=1)

JP_FONT_PATH = None
jp_font = find_japanese_font(unique_char, JP_FONT_PATH)
if jp_font is None:
    print("No compatible Japanese font found: image titles omit character labels.")

pairs = [("エ", "ヱ"), ("ン", "シ"), ("チ", "テ")]
examples_per_row = 4

for actual, predicted in pairs:
    groups = [
        ("Confusion", inspection[
            inspection["Actual"].eq(actual)
            & inspection["Predicted"].eq(predicted)
        ]),
        ("Correct source", inspection[
            inspection["Actual"].eq(actual) & inspection["Correct"]
        ]),
        ("Correct target", inspection[
            inspection["Actual"].eq(predicted) & inspection["Correct"]
        ]),
    ]
    print(f"Actual {actual}, confused with {predicted}: {len(groups[0][1])} errors")
    fig, axes = plt.subplots(3, examples_per_row, figsize=(10, 7), squeeze=False)
    fig.suptitle("Errors and correctly classified reference images")

    for row, (group_name, candidates) in enumerate(groups):
        axes[row, 0].text(
            -0.12, 0.5, group_name, transform=axes[row, 0].transAxes,
            rotation=90, va="center", ha="right",
        )
        # Fixed seed makes the displayed selection reproducible.
        selected = candidates.sample(
            n=min(examples_per_row, len(candidates)), random_state=SEED,
        )
        for ax in axes[row]:
            ax.set_axis_off()
        if selected.empty:
            axes[row, 0].text(0.5, 0.5, "No examples", ha="center")
        for ax, (_, example) in zip(axes[row], selected.iterrows()):
            with Image.open(example["image_path"]) as image:
                ax.imshow(image.convert("L"), cmap="gray", vmin=0, vmax=255)
            if jp_font is not None:
                ax.set_title(
                    f"{example['Actual']} → {example['Predicted']}",
                    fontproperties=jp_font,
                )
            ax.text(
                0.5, -0.08,
                f"Score: {example['confidence']:.1%}\nWriter: {example['writer_id']}",
                transform=ax.transAxes, ha="center", va="top", fontsize=9,
            )
    fig.tight_layout(rect=(0.03, 0, 1, 0.94), h_pad=3)
    plt.show()
```

## Exploratory check: upper-stroke orientation in チ

Sixty validation images of チ were sampled with seed 12. Their upper strokes were visually annotated by the assistant before examining the corresponding predictions. Horizontal means approximately level, inclined means clearly sloping, and ambiguous covers curved, unclear, or borderline cases. These subjective categories use no measured angle threshold and should be reviewed before a stronger conclusion is drawn.

Annotations are saved with image paths in `reports/chi_orientation_annotations.csv`, so they remain linked to the same images even if row order changes. They are exploratory analysis metadata, not training labels. The table below tests association with チ → テ errors, not causality. Small group sizes limit interpretation.

```python
annotations = pd.read_csv(project_root / "reports" / "chi_orientation_annotations.csv")

# Join by image path, not by the current row order.
orientation_results = inspection.copy()
orientation_results["image_path"] = orientation_results["image_path"].map(
    lambda path: str(Path(path).relative_to(project_root))
)
annotated_chi = annotations.merge(
    orientation_results[["image_path", "Actual", "Predicted", "Correct"]],
    on="image_path",
    validate="one_to_one",
)
assert len(annotated_chi) == 60
assert annotated_chi["Actual"].eq("チ").all()
annotated_chi["Confused with テ"] = annotated_chi["Predicted"].eq("テ")
annotated_chi["Any error"] = ~annotated_chi["Correct"]

orientation_comparison = annotated_chi.groupby("orientation").agg(
    Images=("Correct", "size"),
    Confusions_with_テ=("Confused with テ", "sum"),
    Confusion_rate=("Confused with テ", "mean"),
    Overall_error_rate=("Any error", "mean"),
)
display(orientation_comparison.style.format({
    "Confusion_rate": "{:.1%}",
    "Overall_error_rate": "{:.1%}",
}))
```
